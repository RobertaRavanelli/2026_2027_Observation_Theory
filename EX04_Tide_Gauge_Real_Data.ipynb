# **Tide Gauge Data**

In this exercise, you will process 10 years of sea-level data (2006–2016) collected by the tide gauge station in IJmuiden.

The dataset is provided as a text file (available in github): [`Ijmuiden.txt`](https://github.com/RobertaRavanelli/2026_2027_Observation_Theory/blob/main/sample_data/Ijmuiden.txt).

The initial code loads the data and stores:

*   the time epochs (in years) in the variable named `t`;
*   the sea level measurements (in millimeters) in the variable named `y`.

We will consider two different linear models, described by the following observation equations:


 **Model 1 (linear trend model):** $$ \LARGE y_i = l_0 + (t_i - t_0) \cdot r →A1$$

 where:


*   $l_0$ is the initial sea level [mm] at the reference time $t_0$ (in our case $t_0$ is the epoch of the first observation);
*  $r$ is the rate of change of sea level [mm/year].




 **Model 2 (linear trend + annual harmonic):** $$ \LARGE y_i = l_0 + r(t_i- t_0)+ a_c\cos(2\pi (t_i- t_0))  + a_s\sin(2\pi (t_i-t_0))  →A2$$

 where:

 *   $l_0$ is the initial sea level [mm] at the reference time $t_0$ (in our case $t_0$ is the epoch of the first observation);
*  $r$ is the rate of change of sea level [mm/year];
*   $a_c$ is the amplitude of the cosin term [mm] at the reference time $t_0$;
*  $a_s$ is the amplitude of the sin term [mm] at the reference time $t_0$.
