# EX09 - Confidence intervals

In this Python exercise, we will investigate a curve fitting problem using a synthetic sea-level-rise dataset.

The dataset is provided in the [`W6_syntheticdata4.txt`](https://github.com/RobertaRavanelli/2026_2027_Observation_Theory/blob/main/sample_data/W6_syntheticdata4.txt) file available on github. It contains yearly sea level measurements (in cm) collected over 20 years.

The observations epochs, expressed in years, are as `[1, 2, 3, ..., 20]`.

Assume that the observations are independent and normally distributed, with a standard deviation of σ = 2 cm.

The sea level at time zero ($t_0$ = 0), denoted by $x_0$, is unknown and must be estimated together with the other model parameters.

A-priori information:

* there may be at most one outlier in the dataset;

* the correct functional model is uncertain: both a linear trend and a quadratic model (representing an accelerating rise) are plausible;

* the assumed standard deviation, σ = 2 cm, may be too optimistic.
  
For both functional models, complete the following steps:

1. import the dataset and define the observation vector $y$ and the corresponding observation epochs $t$;
2. define the covariance matrix $Q_{yy}$ (stochastic model);
3. define the design matrix $A$ (functional model), assuming:
      * the linear trend model $E\{\underline{y}\} = x_0 + x_1 \cdot t$ (first cell);
      * the quadratic model $E\{\underline{y}\} = x_0 + x_1 \cdot t + x_2 \cdot t^2$ (second cell);
4. apply BLUE to estimte the model parameters: $\hat{x}={(A^T{Q_{yy}^{-1}}A)}^{-1}A^T{Q_{yy}^{-1}}y$;
5. compute the adjusted observation: $\hat{y}= A\hat{x}$;
6. compute the residuals: $\hat{e}= y - \hat{y}$;
7. compute the weighted sum of the squared residuals: $\hat{e}^T{Q_{yy}^{-1}}\hat{e}$;
8. compute the covariance matrix of the estimates: $Q_{\hat{x}\hat{x}}={(A^T{Q_{yy}^{-1}}A)}^{-1}$;
9. compute the covariance matrix of the adjusted onservations: $Q_{\hat{y}\hat{y}}= A Q_{\hat{x}\hat{x}} A^T$;
10. compute the pointwise confidence intervals at 95% probability of the adjusted observations $\hat{y}$, using the using the diagonal elements of the covariance matrix of the adjusted onbervations $Q_{\hat{y}\hat{y}}$:
      * $ {\hat{y}_{i}}_{L} =  \hat{y}_{i} - scaling\_factor(conf\_level) \cdot {\sigma}_{\hat{y}_{i}} $;
      * $ {\hat{y}_{i}}_{U} =  \hat{y}_{i} + scaling\_factor(conf\_level) \cdot {\sigma}_{\hat{y}_{i}} $;
11. identify the observation with the largest absolute residual .
