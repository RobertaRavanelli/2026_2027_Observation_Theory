# EX 06 - Best Linear Unbiased Estimator (BLUE)

A ballistic body is an object moving through the air after being launched, with its motion governed primarily by gravity. In the idealized model considered here, air resistance is neglected, so the vertical acceleration is assumed to be constant.

Under this assumption, the body's altitude varies quadratically with time. By estimating the coefficients of this quadratic model, we can determine the body's initial altitude, initial vertical velocity, and vertical acceleration.

To estimate these motion parameters, eight altitude observations were collected at different times along the body's trajectory.



Schermata 2020-10-29 alle 23.19.38.png

The observations $y_i$ at each epoch $t_i$ are given below:


*   Time `[seconds]`:  `[1,2,3,4,5,6,7,8]`
*   Observed altitude `[meters]`:  `[79.06,126.13,179.90,218.77,258.74,266.91,274.91,280.61]`

In the simplest case (air resistance neglected), the altitude of the ballistic body can be modeled as a quadratic function of time:

Model:  

$$ \large y_i = x_0 + x_1t_i + \frac{1}{2} x_2t_i^2$$

where:

*   $𝑥_0$  is the initial altitude [m]
*   $𝑥_1$  is the initial vertical velocity [m/s]
*   $𝑥_2$  is the vertiacla accelleration (gravity acceleration) [m/s²]

The precision of the observations varies (improves) with time. The variance of each observation is given by

$$ \large \sigma_{y_i}^2 = \frac{1}{10\sqrt{t_i}} \; \ \ \ \ \ \      [m^2]$$

The observations are assumed to be uncorrelated.

Estimate the model parameters using the Best Linear Unbiased Estimator (BLUE) and assess the estimated vertical acceleration, $𝑥_2$.
